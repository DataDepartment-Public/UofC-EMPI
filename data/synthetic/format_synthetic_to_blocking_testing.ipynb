```python
import pandas as pd
```

```python
df = pd.read_csv('synthetic_test_v3.csv', dtype=str)
```

```python
left_record_columns = ['PATID_A'] + [c for c in df.columns if c.endswith('_l')]
right_record_columns = ['PATID_B'] + [c for c in df.columns if c.endswith('_r')]
```

```python
df_concat = pd.concat([
    df[left_record_columns].rename(columns=lambda x: x[:-2]),
    df[right_record_columns].rename(columns=lambda x: x[:-2])
], ignore_index=True)
df_concat.drop_duplicates(inplace=True)
```

```python
df_concat.to_csv('synthetic_blocking_testing.csv', index=False)
```
